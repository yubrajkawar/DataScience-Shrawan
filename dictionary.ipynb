# Dictionary Comphrension
-concise way to make or create the dictionary in short and clear way.

Syntax:
{key:value for (key,value) in iterable if condition}



# Creating a dictionary from two list
zip() - function merge two iterable items (lists, sets, dictionary, and so on)

```python
keys=['a','b','c','d','e'] #list of keys
value=[1,2,3,4,5] # list of values
dict = {k:v for (k,v) in zip(keys,value)}   # we are using zip() function to combine key and value respectively
print(dict)
```

```python
pairs=[('apple',1),('banana',2),('orange',3)] # tuple inside list
dict={fruit: number for (fruit,number) in pairs}  # since there aren't 2 lists so we don't use zip () function to combine
print(dict)
print(type(dict))
```

```python
a = [("a",1),("b",2),("c",3)]
b=[('x',4),('y',5),('z',6)]
new_d = {
    key:value for (key,_),(_,value)in zip(a,b)
    
}
print(new_d)
```

```python
#create a dictionary using dictionary comprehension where the numbers 1 to 10 are the keys and their squares are the values.
dict={key:key**2 for key in range(1,11)}
print(dict)
```
