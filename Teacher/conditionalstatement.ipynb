Write a python program that ask user for age , monthly salary and credit score
Rules:
1.The customers must be at least 21 years old
2.salary must be at least Rs.30,000
3.Credit score must be 700 or higher
4.If all conditions are satisified , print("Loan Approved")
5.print("Loan rejecteed")

```python
age = int(input("Enter your age: "))
salary = float(input("Enter your salary: "))
credit_score = int(input("Enter your credit score: "))

if age >=21:
    if salary >=30000:
        if credit_score >=700:
            print("Loan Approved")
        else:
            print("Loan Denied: Credit score too low")
    else:
        print("Loan Denied: Salary too low")
else:
    print("Loan Denied: Age too low")
```

Write a python program that ask the customer for :
Total shooping amount and wwhether they are memeber (yes or no)
rules:
-if the amount is 5000 or more and the customer is a memeber - 25%discount
-otherwise  if the amount is 5000 or more - 10%discount 
otherwise -no discount 
pRint- discount amount and the final amount to pay 
|

```python
amount = float(input("Enter shopping amount: "))
member = input("Are you a member? (yes/no): ")

if amount >= 5000 and member == "yes":
    discount = amount * 0.20
else:
    if amount >= 5000:
        discount = amount * 0.10
    else:
        discount = 0

final_amount = amount - discount

print("Discount:", discount)
print("Final amount:", final_amount)
```
