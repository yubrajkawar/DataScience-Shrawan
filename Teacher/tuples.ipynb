# Python dictionaries :
-it is collection of an items , similar to list and tuples
-it consists of key values pair.
-it is created by placing key/value pairs using { }-curly braces

```python
person = {
    "name":"John",
    "age":30,
    "address":"123 Main St"
}
print(person)
print(person['name'])
person['phone']=555555555555
print(person)
```

Assign the dictionary below to a variable named prices
{apple -2.5,kiwi-3.4}
then , change the value of apple to 3.5 and add an item with key banana and value 3 to the dictionary and print the dictionary.

```python
prices = {
    'apple':2.5,
    'kiwi':3.4
}
print(prices)
prices['apple']=3.5
print(prices)
prices['banana']=3
print(prices)
```

```python
students ={
    "student1":{
        "name":"John",
        "age":20,
        "marks":85
    },
    "student2":{
        "name":"Alice",
        "age":22,
        "marks":90
    }
}
print(students)
```

```python
students = {
    'name':['John','Alice'],
    'age':[20,22],
    'course':['Math','Science'],
    'marks':[85,90]
}
print(students)
print(students['name'][0])
```

```python
numbers = {1,2,3,4,5}
numbers.add(6)
print(numbers)
```
