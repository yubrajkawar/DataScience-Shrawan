What is a Class?

Think of a class as a blueprint or recipe for creating something. For example, if you have a blueprint for building a house, you can use it to build many houses. Each house can have its own color or owner, but they all follow the same basic plan.

In Python, a class defines what properties (data) and actions (functions) its objects will have.

An object is like the actual house built from the blueprint. It is a real thing created using the class. Each object can have its own unique values, but it is made using the class as a guide.

In Python, when you create an object from a class, it is called an instance of that class.

-----------------------------------------------------------------------------
Simple Analogy
Class: The blueprint for a car.

Object: A specific car made from that blueprint, like your red Toyota or your friend's blue Honda.

## core principles:
Encapsulation: Encapsulation is the bundling of data (attributes) and methods (functions) that operate on that data into a single unit, i.e., a class. It restricts direct access to some of the object’s components, which is typically done by making attributes private and accessing them through public getter and setter methods. This ensures data security and hides implementation details from the user.
Inheritance: Inheritance allows one class (child) to acquire the properties and behaviors of another class (parent). This promotes code reuse and creates a natural hierarchy of classes. The child class can override or extend the functionality of the parent class.
Polymorphism: The ability of an object to take on many forms. This often involves methods with the same name but different implementations (e.g., a draw() method that behaves differently for a Circle object vs. a Square object).
Abstraction: Abstraction involves hiding the complex implementation details and showing only the essential features of an object. For example, using a smartphone you tap an app icon, not the underlying operating system code.

```python
# self is a reference to the current instance of a class and using it you access and modify the object’s 
# attributes from within the class.

# It’s used inside a class’s methods to refer to the specific object the method is being called on. 
# When you call a method on an object, Python implicitly passes the object itself as the first 
# argument to that method. We conventionally name this first argument self.

# For example, if you have a Dog class and you create a my_dog object, when you call my_dog.bark(), 
# Python internally translates this to something like Dog.bark(my_dog). So, self inside the bark() method becomes a reference to my_dog. It’s how the bark() method knows which dog to bark (in case you have multiple dog objects).

# Also, using self, we can then access and modify the attributes of that specific dog. 
# For example, inside the bark() method, we might want to print the dog’s name: print(f"{self.name} is barking!"). self.name refers to the name 
# attribute of the particular dog object the bark() method is called on.
```

```python
class Dog:
    def __init__(self, name):
        self.name = name  # self.name is the object's attribute, name is the argument

    def bark(self):
        print(f"{self.name} is barking!") # self.name refers to the specific dog's name

dog1 = Dog("Buddy")
dog2 = Dog("Max")

dog1.bark()  # Output: Buddy is barking!
dog2.bark()  # Output: Max is barking!
```

Create a Python program using object-oriented programming that allows users to track student marks, calculate total and average scores, and assign letter grades based on performance.
---------------------------------------------------------------------------------------------------

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks  # a list of marks for subjects
        self.total = 0
        self.average = 0
        self.grade = ""

    def calculate_total_average(self):
        self.total = sum(self.marks)
        self.average = self.total / len(self.marks)

    def assign_grade(self):
        if self.average >= 90:
            self.grade = 'A'
        elif self.average >= 80:
            self.grade = 'B'
        elif self.average >= 70:
            self.grade = 'C'
        elif self.average >= 60:
            self.grade = 'D'
        else:
            self.grade = 'F'

    def show_result(self):
        print(f"\nStudent Name: {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Total Marks: {self.total}")
        print(f"Average Marks: {self.average:.2f}")
        print(f"Grade: {self.grade}")


# Main program
name = input("Enter student name: ")
marks = []

# Take input for 3 subjects
for i in range(1, 4):
    mark = float(input(f"Enter marks for Subject {i}: "))
    marks.append(mark)

# Create object and process data
student = Student(name, marks)
student.calculate_total_average()
student.assign_grade()
student.show_result()
```

Find the factorial of a number" using OOP concept


```python
class FactorialCalculator:
    def __init__(self, number):
        self.number = number

    def calculate_factorial(self):
        if self.number < 0:
            return "Factorial is not defined for negative numbers."
        factorial = 1
        for i in range(1, self.number + 1):
            factorial *= i
        return f"Factorial of {self.number} is {factorial}"

# Example usage
num = int(input("Enter a number: "))
fact_obj = FactorialCalculator(num)
print(fact_obj.calculate_factorial())
```

Q: Write a Python program to implement a BankAccount class with the following features:

1.Constructor (__init__)

    Accepts account_holder, account_number, and an optional balance (default is 0).

    Initializes instance variables.

2.Method: deposit(amount)

    Adds the amount to the balance if the amount is greater than 0.

    Returns a message confirming the deposit and updated balance.

3.Method: withdraw(amount)

    Deducts the amount from balance if it's less than or equal to the current balance.

    Returns a message confirming the withdrawal and updated balance or a message for insufficient funds.

4.Method: get_balance()

    Returns a message showing the current balance.

5.Write sample usage code:

Create an object account of BankAccount class with initial balance of $1000.

Print the account balance.

Deposit $500 and show the result.

Withdraw $200 and show the result.

```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            return f"Deposited ${amount}. New balance: ${self.balance}"
        return "Invalid deposit amount"

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            return f"Withdrew ${amount}. New balance: ${self.balance}"
        return "Invalid or insufficient funds"

    def get_balance(self):
        return f"Account balance: ${self.balance}"

# Example usage
account = BankAccount("John Doe", "1234567890", 1000)
print(account.get_balance())  
print(account.deposit(500))   
print(account.withdraw(200))  
```

```python
#lambda sorting in class:
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
students = [
    Student("Alice", 88),
    Student("Bob", 75),
    Student("Charlie", 92)
]

# Sort students by grade using lambda
print(students.sort(key=lambda self: self.grade))  #The sort() method sorts the students list in-place.
# key=lambda s: s.grade tells sort to use each student's grade attribute as the sorting key.

# lambda s: s.grade is an anonymous function that takes a Student object s and returns its grade.

for item in students:
    print(item.name, item.grade)
```

```python
from datetime import datetime, timedelta
from openpyxl import Workbook
from openpyxl.styles import Font

class Book:
    def __init__(self, title, author, isbn, total_copies):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.total_copies = total_copies
        self.available_copies = total_copies
    
    def __str__(self):
        return f"'{self.title}' by {self.author} (ISBN: {self.isbn}) [Available: {self.available_copies}/{self.total_copies}]"

class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []  # Stores ISBNs
    
    def __str__(self):
        return f"{self.name} (ID: {self.member_id}) - Borrowed: {len(self.borrowed_books)} books"

class Library:
    def __init__(self):
        self.books = []
        self.members = []
        self.transactions = []
    
    def add_book(self, title, author, isbn, copies):
        self.books.append(Book(title, author, isbn, copies))
        return f"Added '{title}' ({copies} copies)"
    
    def add_member(self, name, member_id):
        self.members.append(Member(name, member_id))
        return f"Added member: {name}"
    
    def borrow_book(self, member_id, isbn):
        member = next((m for m in self.members if m.member_id == member_id), None)
        book = next((b for b in self.books if b.isbn == isbn), None)
        
        if not member or not book:
            return "Member or book not found!"
        if book.available_copies <= 0:
            return "No copies available!"
        
        book.available_copies -= 1
        member.borrowed_books.append(isbn)
        due_date = datetime.now() + timedelta(days=14)
        self.transactions.append({
            "member_id": member_id,
            "isbn": isbn,
            "date": datetime.now().strftime("%Y-%m-%d"),
            "due_date": due_date.strftime("%Y-%m-%d"),
            "returned": False
        })
        return f"'{book.title}' borrowed. Due: {due_date.strftime('%Y-%m-%d')}"
    
    def export_to_excel(self, filename="library_report.xlsx"):
        wb = Workbook()
        
        # Books sheet
        ws_books = wb.active
        ws_books.title = "Books"
        ws_books.append(["Title", "Author", "ISBN", "Total Copies", "Available"])
        for book in self.books:
            ws_books.append([book.title, book.author, book.isbn, book.total_copies, book.available_copies])
        
        # Members sheet
        ws_members = wb.create_sheet("Members")
        ws_members.append(["Name", "Member ID", "Books Borrowed"])
        for member in self.members:
            ws_members.append([member.name, member.member_id, ", ".join(member.borrowed_books)])
        
        # Transactions sheet
        ws_trans = wb.create_sheet("Transactions")
        ws_trans.append(["Member ID", "ISBN", "Borrow Date", "Due Date", "Returned"])
        for trans in self.transactions:
            ws_trans.append([
                trans["member_id"],
                trans["isbn"],
                trans["date"],
                trans["due_date"],
                "Yes" if trans["returned"] else "No"
            ])
        
        # Style headers
        for sheet in wb:
            for cell in sheet[1]:
                cell.font = Font(bold=True)
        
        wb.save(filename)
        return f"Data exported to {filename}"

def main():
    lib = Library()
    
    # Sample data
    lib.add_book("Python Crash Course", "Eric Matthes", "978-1593279288", 5)
    lib.add_book("Clean Code", "Robert Martin", "978-0132350884", 3)
    lib.add_member("Alice Johnson", "M001")
    lib.add_member("Bob Smith", "M002")
    
    # Sample transactions
    lib.borrow_book("M001", "978-1593279288")
    lib.borrow_book("M002", "978-0132350884")
    
    while True:
        print("\n==== Library Menu ====")
        print("1. List all books")
        print("2. List all members")
        print("3. Borrow a book")
        print("4. Return a book")
        print("5. Export to Excel")
        print("6. Exit")
        
        choice = input("Enter choice (1-6): ")
        
        if choice == "1":
            print("\nAll Books:")
            for book in lib.books:
                print(book)
        
        elif choice == "2":
            print("\nAll Members:")
            for member in lib.members:
                print(member)
        
        elif choice == "3":
            member_id = input("Member ID: ")
            isbn = input("Book ISBN: ")
            print(lib.borrow_book(member_id, isbn))
        
        elif choice == "4":
            member_id = input("Member ID: ")
            isbn = input("Book ISBN: ")
            # Implement return logic here
            print("Return functionality would go here")
        
        elif choice == "5":
            filename = input("Excel filename (default: library_report.xlsx): ") or "library_report.xlsx"
            print(lib.export_to_excel(filename))
        
        elif choice == "6":
            print("Goodbye!")
            break
        
        else:
            print("Invalid choice!")

if __name__ == "__main__":
    main()
```

```python
class Library:
    def __init__(self, book_list):
        self.books = book_list

    def display_books(self):
        print("\nAvailable Books:")
        for book in self.books:
            print(f" - {book}")

    def borrow_book(self, book_name):
        if book_name in self.books:
            self.books.remove(book_name)
            return f"You have borrowed '{book_name}'. Please return it within 7 days."
        else:
            return f"'{book_name}' is not available in the library."

    def return_book(self, book_name):
        self.books.append(book_name)
        return f"Thank you for returning '{book_name}'."

# Example usage
library = Library(["Python Programming", "Data Structures", "Artificial Intelligence", "Machine Learning"])

while True:
    print("\n--- Library Menu ---")
    print("1. Display Books")
    print("2. Borrow Book")
    print("3. Return Book")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        library.display_books()
    elif choice == '2':
        book = input("Enter the book name you want to borrow: ")
        print(library.borrow_book(book))
    elif choice == '3':
        book = input("Enter the book name you want to return: ")
        print(library.return_book(book))
    elif choice == '4':
        print("Thanks for using the library system. Goodbye!")
        break
    else:
        print("Invalid choice. Please select from 1 to 4.")
```

#                                                    Polymorphism in OOP PYthon:  ------------------

Polymorphism in OOP is the ability of an object to take many forms. In simple words, polymorphism allows us to perform the same action in many different ways.

Polymorphism is taken from the Greek words Poly (many) and morphism (forms). Polymorphism defines the ability to take different forms.


It allows the same function or method name to behave differently based on the object (or data type) it's acting on.

```python
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

# Polymorphic function
def make_sound(self):
    self.sound()

# Create objects
d = Dog()
c = Cat()

# Call the same function with different types
make_sound(d)
make_sound(c)
```

```python

class Bird:
    def speak(self):
        print("Some sound")

# Step 3: Define Parrot class inheriting from Bird
class Parrot(Bird):
    def speak(self):
        print("Parrot says hello")

# Step 4: Define Crow class inheriting from Bird
class Crow(Bird):
    def speak(self):
        print("Crow says caw")

# Step 5: Define make_sound function
def make_sound(bird):
    bird.speak()

# Step 6: Create Parrot object
p = Parrot()

# Step 7: Create Crow object
c = Crow()

# Step 8: Call make_sound with Parrot
make_sound(p)  # Output: Parrot says hello

# Step 9: Call make_sound with Crow
make_sound(c)  # Output: Crow says caw

```
