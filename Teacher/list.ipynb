```python
#deleting an item from the list
animals = ['cat', 'dog', 'rabbit', 'hamster']
del animals[2]
print(animals)  # Output: ['cat', 'dog', 'hamster']
del animals 
print(animals)  # This will raise a NameError since the list has been deleted
```

```python
animals = ['cat', 'dog', 'rabbit', 'hamster']
animals.append('parrot')
print(animals)
animals[1]='fish'
print(animals)
animals.insert(2,'tiger')#insert an item at a specific index pushing the indexes of the other items to the right
print(animals)
```
