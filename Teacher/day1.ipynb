# Explicit Type Conversion:
-also called type casting 
user manually changes the value from one data type to a another data type 
-here user defined its own data type to change  

# Common type of type casting functions:
1.int()-convert a value to an integer
2.float()-convert a value to floating numbers
3.str()-convert a value to the string
4.bool()-convert a value to boolean (True /False)


```python
float_number = 10.656
integer = int(float_number) #explicit type conversion
print(integer)
```

```python
string = "10"
print(type(string))
integer = int(string) #explicit type conversion
print(integer)
print(type(integer))
```

```python
#Converting a binary string to an integer
s = "1101"
a =int(s,2) # a = int(s,2 is base of integer)
print(a) 
```
