iloc is used to select rows and columns by their integer position (index number).
Syntax:
df.iloc[row, column]
For instance:
df.iloc[0:4, 0:3]
This selects:

Rows 0 to 3
Columns 0 to 2

Differences:
loc → labels / conditions
iloc → integer positions

| Point                   | `loc`                           | `iloc`                              |
| ----------------------- | ------------------------------- | ----------------------------------- |
| **1. Selection method** | Selects using **labels/names**  | Selects using **integer positions** |
| **2. Syntax**           | `df.loc[row, column]`           | `df.iloc[row, column]`              |
| **3. Example**          | `df.loc[0:3, ['name', 'math']]` | `df.iloc[0:4, 0:2]`                 |

This code filters students based on two conditions.

`&` means **AND**, so both conditions must be true.
Therefore, it returns students with **attendance ≥ 85 and average score ≥ 80**.


```python
df[(df['attendance'] >= 85) & (df['average_score'] >= 80)]
```

```python
df[(df['city'] == 'Kathmandu') | (df['city'] == 'Pokhara')]
```

## 13. `isin()`
isin() is used to check whether values in a column belong to a specific list of values.
Syntax:
df[df['column_name'].isin([value1, value2, value3])]


For example, if you want to find students from Kathmandu or Pokhara:
df[df['city'].isin(['Kathmandu', 'Pokhara'])]

df['city'] → selects the city column.
.isin(['Kathmandu', 'Pokhara']) → checks whether each city is Kathmandu or Pokhara.
df[...] → returns only the students who match.

“Give me rows where the value is one of these specified values.”

```python
df[df['city'].isin(['Kathmandu', 'Pokhara'])]
```

```python
df[df['department'].isin(['IT', 'Science'])]
```

## 14. `between()`

This is why because of this:

between() = filter values within a range.

```python
df[df['age'].between(20, 22)]
```

## 18. `np.where()` for categories
np.where() is used to create a new column based on a condition. It allows us to assign one value when the condition is True and another value when it is False.
np.where(condition, value_if_true, value_if_false)

df['performance'] = np.where(df['average_score'] >= 80, 'Excellent', 'Needs Improvement')
This checks the average_score of each student:

If score ≥ 80 → "Excellent"
If score < 80 → "Needs Improvement"

np.where() = if condition is true, give one value; otherwise, give another value.

| Feature            | `df['pass_status'] = df['average_score'] >= 60` | `np.where()`                                    |
| ------------------ | ----------------------------------------------- | ----------------------------------------------- |
| **Result**         | Creates `True` or `False`                       | Creates custom values like `"Pass"` or `"Fail"` |
| **Example output** | `True`, `False`, `True`                         | `"Pass"`, `"Fail"`, `"Pass"`                    |
| **Use**            | When you only need a **Boolean condition**      | When you want **specific categories/labels**    |


```python
import numpy as np
```

```python
df['performance'] = np.where(df['average_score'] >= 80, 'Good', 'Needs Improvement')
```

```python
df[['name', 'average_score', 'performance']]
```

## 23. Sort by index:
Not exactly. sort_index() is for sorting the DataFrame by its index, which is usually the row labels, not for sorting the values inside a row.
df.sort_index()

| Function                  | Sorts by                    |
| ------------------------- | --------------------------- |
| `sort_values()`           | Values in a **column**      |
| `sort_index()`            | **Row/column index labels** |
| `df.loc[0].sort_values()` | Values **inside row 0**     |
