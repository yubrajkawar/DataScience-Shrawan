# Boolean indexin: 

It allows us to select and modify elements of an array based on boolean conditoins.

```python
numbers=np.array([10,20,30,40,55,60,77,80])
even=numbers[numbers % 2 == 0]   # [condition], first check list of True or False and prints only the True ones
print(even)
```

```python
salaries=np.array([1000,2000,3000,4000,5000])
print(salaries)
outlier_mask=salaries>1500  #result of condition is Boolean "True" or "false"
print(outlier_mask)
outliers=salaries[outlier_mask] #Otulier_masks is true and false and print only true ones
print(outliers)
```

```python
predict=np.array([0.1,0.85,0.3,0.92,0.45,0.78,0.05])
greater=predict>0.5
print(greater)   #print Booleaan of True or false list
predict[greater]=1.0
predict[~greater]=0.0
print("Binary Predictions",predict)
```

# slicing - taking a part of an array

Syntax:
array[start:stop:step]


: -slicing operator

```python
import numpy as np

numbers=np.array([10,20,30,40,50,60,70,80])
sliced_number=numbers[:3] #start empty and end specified so 0, 1, 2, and 3
print(sliced_number)
```

Write a program to slice NumpY array :
1.take the numpy array  numbers as user input 
2.slice the array numbers from the index 1 to 4
3.slice the array numbers from index 2 to the last elements
4.slice the array numbers from the first to last without using any numbers
5.print the sliced arrays.

test input:[1 ,2 ,3, 4, 5]

```python
import numpy as np

numbers=np.array(list(map(int,input("Enter the numbers:").split())))
print(numbers)
sliced_numbers=numbers[1:4]
print(sliced_numbers)
slice_2nd_last=numbers[2:]
print(slice_2nd_last)
slice_1st_last=numbers[: :]
print(slice_1st_last)
```

For 3D array slicing:

Syntax:
array[block,row, column]

1st layer as index 0 in array[0]

```python
# For 3D-array
np_array=np.array(
    [
        [
        [1,2,3],
        [4,5,6],
        [7,7,8]
        ],
        [
            [9,10,11],
            [12,14,15],
            [15,16,17]
        ]
    ]
)
print(np_array[1])  #print 1st block 
print(np_array[1,0,2]) #print block 1's, 0 row's, 2nd column's
print(np_array[:,:,:]) #Take all blocks, all rows, all columns
```

```python
blocks =int(input("Enter the number of blocks:"))
rows = int(input("Enter the number of rows:"))
cols = int(input("Enter the number of columns:"))
arr =[]
for i in range(blocks):
    block =[]
    for j in range(rows):
        row = list(map(int,input(f"enter {cols} values for block{i},row{j}").split()))
        block.append(row)
    arr.append(block)
arr = np.array(arr)
print(arr)
```
