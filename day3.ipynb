#Data Conversion: Type Conversion
Process of changing the value from one data type to another data types.
from float to int, int to str

There are 2 types:
1. implicit conversion 
2. Explicit conversion

1. Implicit type conversion: That happens automatically. No data loss

```python
# This is an example of implicit data conversion.
x=10
y=25.6
add=x+y
print(add)
# Here, data is automatically converted to float type in variable "add" without defining ourselves.
```

2) Explicit Type COnversion
THis is also called Type Casting. 
Here, user manually changes the value from one data type to another.

Common Types of Type Casting:
a. int()- convert a value to an integer.
b. float()- convert a value to floating numbers.
c. str() - convert a value to the string
d. bool()- convert a value to boolean (True / False)

```python
def analyze(numbers):
    even = []
    odd = []
    for number in numbers:
        if number % 2 ==0:
            even.append(number)
        else:
            odd.append(number)
    largest = max(numbers)
    total = sum(numbers)
    return even ,odd,largest ,total

numbers =[]
n = int(input("eneter how many the number that you want to enter:"))
for i in range(n):
    number = int(input("enter number"))
    numbers.append(number)

even,odd,largest,total = analyze(numbers)
print("even numbers:",even)
print("odd numbers",odd)
print("largest numbers ",largest)
print("total",total )
```

```python
class Student:  # use generally capital letter for making class
    #init method
    def __init__(self,name,score):
        self.name=name      # saves a value to specific object so the object can remember
        self.score=score    # we initialized the method

    def check_pass_fail(self):
        if self.score>=40:
            return True
        else:
            return False

    student1=Student("Nishan",85)
    passorfail=student1.check_pass_fail()
    print(f"{student1.name} pass",passorfail)
    student2=Student("Ram",35)
    passorfail=student2.check_pass_fail()
        
```

```python
class Student:                         # Create a class named Student

    def __init__(self, name, score):   # __init__ runs automatically when an object is created
        self.name = name               # Store name inside this specific object
        self.score = score             # Store score inside this specific object


    def check_pass_fail(self):         # Create a method to check whether student passed
        if self.score >= 40:           # Check the score of the current object
            return True                # If score is 40 or more, return True
        else:
            return False               # If score is below 40, return False


student1 = Student("Nishan", 85)       # Create student1; name="Nishan", score=85

passorfail = student1.check_pass_fail() # Call method for student1
                                        # self automatically becomes student1
                                        # Returns True because 85 >= 40

print(f"{student1.name} pass", passorfail) # student1.name is "Nishan"
                                           # Prints: Nishan pass True


student2 = Student("Ram", 35)          # Create student2; name="Ram", score=35

passorfail = student2.check_pass_fail() # Call method for student2
                                        # self automatically becomes student2
                                        # Returns False because 35 < 40

print(f"{student2.name} pass", passorfail) # Prints: Ram pass False
```

```python
class Student:                              # Create a class named Student

    def __init__(self, name, age, marks):   # Constructor; runs automatically when object is created
        self.name = name                    # Store name in the student object
        self.age = age                      # Store age in the student object
        self.marks = marks                  # Store marks list in the student object
    def average(self):                      # Create a method to calculate average
        return sum(self.marks) / len(self.marks)
                                            # sum() adds all marks
                                            # len() tells how many marks there are
                                            # return sends the average back
    
    def display(self):                      # Create a method to display student information
        print("Name:", self.name)           # Print the student's name
        print("Age:", self.age)             # Print the student's age
        print("Marks:", self.marks)         # Print the student's marks
        print("Average:", self.average())   # Call average() and print the result

name = input("Enter the name: ")            # Ask user for name
age = int(input("Enter the age: "))         # Ask for age and convert input to integer
marks = []                                  # Create an empty list to store marks
for i in range(3):                          # Repeat 3 times to get 3 marks
    mark = float(input(f"Enter the marks {i+1}: "))
                                            # Ask user to enter a mark
                                            # float() converts it to a decimal number
    marks.append(mark)                      # Add the mark to the marks list
student = Student(name, age, marks)         # Create a Student object
                                            # name, age and marks are passed to __init__()
student.display()                           # Call display() method
```

Q: Write a Python program to implement a BankAccount class with the following features:

1.Constructor (_init_)

    Accepts account_holder, account_number, and an optional balance (default is 0).

    Initializes instance variables.

2.Method: deposit(amount)

    Adds the amount to the balance if the amount is greater than 0.

    Returns a message confirming the deposit and updated balance.

3.Method: withdraw(amount)

    Deducts the amount from balance if it's less than or equal to the current balance.

    Returns a message confirming the withdrawal and updated balance or a message for insufficient funds.

4.Method: get_balance()

    Returns a message showing the current balance.

5.Write sample usage code:

Create an object account of BankAccount class with initial balance of $1000.

Print the account balance.

Deposit $500 and show the result.

Withdraw $200 and show the result.

```python
class BankAccount:                                      # Create BankAccount class

    def __init__(self, account_holder, account_number, balance=0):  # "Constructor"; balance defaults to 0
        self.account_holder = account_holder             # Store account holder
        self.account_number = account_number             # Store account number
        self.balance = balance                           # Store account balance

    def deposit(self, amount):                           # "Method" create to deposit money
        if amount > 0:                                   # Check if amount is positive
            self.balance += amount                       # Add amount to balance
            return f"Deposit successful. New balance: ${self.balance}"  # Return deposit message
        else:
            return "Deposit amount must be greater than 0."              # Return error message

    def withdraw(self, amount):                          # "Method" created to withdraw money
        if amount <= self.balance:                       # Check if enough balance exists
            self.balance -= amount                       # Subtract amount from balance
            return f"Withdrawal successful. New balance: ${self.balance}" # Return withdrawal message
        else:
            return "Insufficient funds."                 # Return insufficient funds message

    def get_balance(self):                               # "Method" created to check balance
        return f"Current balance: ${self.balance}"       # Return current balance


account = BankAccount("Nishan", "12345", 1000)           # Create account with $1000 initial balance

print(account.get_balance())                             # Display current balance

print(account.deposit(500))                              # Deposit $500 and display result

print(account.withdraw(200))                             # Withdraw $200 and display result
```

```python
class Person: 
    def __init__(self,name, age):
        self.name=name
        self.age=age
    def display(self):
        print(f"name is: {self.name}")
        print(f"age: {self.age}")

class student(Person):
    def __init__(self,name,age,roll_no,marks):
        super().__init__(name,age)
        self.roll_no=roll_no
        self.marks=marks

    def display(self):
        super().display()
        print(f"RollNo: {self.roll_no}")
        print(f"Marks{self.marks}")

s=student("riya",20,101,89.23)
s.display()
```
