## 7. Making new columns

```python
# price per unit area – a handy derived metric
df['price_per_area'] = df['price'] / df['area']   # op 24: create a new column from arithmetic
df[['price', 'area', 'price_per_area']].head()
```

```python
# tag each house cheap / mid / expensive using cut
df['price_band'] = pd.cut(df['price'],
                          bins=[0, 4000000, 8000000, df['price'].max()],
                          labels=['cheap', 'mid', 'expensive'])   # op 25: pd.cut into bins
df[['price', 'price_band']].head()
```

```python
# map the yes/no text to 1/0 so it's numeric
df['mainroad_flag'] = df['mainroad'].map({'yes': 1, 'no': 0})   # op 26: map() with a dict
df[['mainroad', 'mainroad_flag']].head()
```

```python
# apply a custom function row-wise on one column
df['area_k'] = df['area'].apply(lambda x: round(x / 1000, 1))   # op 27: apply() with lambda
df[['area', 'area_k']].head()
```

## 8. Grouping and aggregatingThe fun part – summarising by category.

```python
df.groupby('furnishingstatus')['price'].mean()   # op 28: mean price per furnishing type
```

```python
# several stats at once per group
df.groupby('bedrooms')['price'].agg(['count', 'mean', 'max'])   # op 29: agg with a list
```

```python
# two-level grouping
df.groupby(['furnishingstatus', 'airconditioning'])['price'].mean()   # op 30: groupby two columns
```

```python
df['furnishingstatus'].value_counts()   # op 31: how many of each category
```

```python
# pivot table: avg price across bedrooms x stories
df.pivot_table(values='price', index='bedrooms',
               columns='stories', aggfunc='mean')   # op 32: pivot_table
```

```python
df['price'].max(), df['price'].min(), df['price'].mean()   # op 33: quick column stats
```

```python
df['area'].corr(df['price'])   # op 34: correlation between area and price
```
